```python
import User.utils.dataloader as dl

optd = dl.Datasets()
train, test = optd('optdigits', preprocess=True)
x = train['data'].reshape(-1, 1)*255
y = train['target'].reshape(-1, 10)
```

```python
print(x)
```

```python
import matplotlib.pyplot as plt
import numpy as np
import pandas as pd


# 绘制直方图，指定区间数（bins）
plt.figure(figsize=(10, 6))
# 保存hist的返回值
counts, bin_edges, patches = plt.hist(x, bins=16, color='skyblue', edgecolor='black', density=True)
plt.title('直方图示例 (16个区间)')
plt.xlabel('数值')
plt.ylabel('概率密度')
plt.grid(True, linestyle='--')
plt.show()

# 将统计结果保存为CSV文件
# 创建一个包含区间和对应频率的DataFrame
hist_data = pd.DataFrame({
    '区间下限': bin_edges[:-1],
    '区间上限': bin_edges[1:],
    '概率密度': counts
})

# 保存到CSV文件
hist_data.to_csv(r'f:\Desktop\temp\python\2D1R_PP\pythonProject\data\histogram_results.csv', index=False)
print("直方图统计结果已保存到 'histogram_results.csv'")

# 如果需要，也可以保存为numpy数组
np.savez(r'f:\Desktop\temp\python\2D1R_PP\pythonProject\data\histogram_data.npz', 
         counts=counts, bin_edges=bin_edges)
print("直方图原始数据已保存到 'histogram_data.npz'")
```
